# Silver Layer Engineering Standards

## Purpose
Create reusable, clinically interpretable and validated MIMIC-IV entities.

## Processing principles
1. Preserve source clinical meaning.
2. Preserve missing values unless a deterministic correction exists.
3. Do not silently remove unusual clinical records.
4. Standardise reusable datatypes, whitespace and units.
5. Validate subject_id, hadm_id, stay_id and domain-specific keys.
6. Enrich coded values using authoritative MIMIC dictionaries.
7. Flag questionable values rather than guessing corrections.
8. Preserve event timestamps.
9. Keep project-specific cohorts and analytical features out of Silver.
10. Record and validate every transformation.
